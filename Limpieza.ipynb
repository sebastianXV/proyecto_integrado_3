# Limpieza y Transformación de Datos

```python
# Importar librerías necesarias
import pandas as pd
from scipy.stats.mstats import winsorize
import numpy as np

# Configuración de visualización
pd.options.display.float_format = '{:,.2f}'.format
pd.set_option('display.max_columns', None)

# URL del dataset
DATASET_URL = 'https://raw.githubusercontent.com/sebastianXV/proyecto_integrado_3/refs/heads/main/Sleep_health_and_lifestyle_dataset.csv'

# Cargar el dataset
try:
    df = pd.read_csv(DATASET_URL)
    print("Dataset cargado exitosamente.")
except Exception as e:
    print(f"Error al cargar el dataset: {e}")

# Eliminar 'Person ID'
if 'Person ID' in df.columns:
    df.drop('Person ID', axis=1, inplace=True)
    print("Columna 'Person ID' eliminada.")

print("\n--- Vista Previa ---")
print(df.head())
```

# 1. Eliminación de Duplicados

```python
# 1. Eliminar filas duplicadas
filas_iniciales = df.shape[0]
df.drop_duplicates(inplace=True)
filas_finales = df.shape[0]

print(f"\nFilas eliminadas por duplicados: {filas_iniciales - filas_finales}")
print(f"Total de filas después de eliminar duplicados: {df.shape[0]}")
```

# 2. Tratamiento de Valores Nulos

```python
# 2. Verificación de Nulos ()
if df.isnull().sum().sum() == 0:
    print("No se requiere tratamiento de valores nulos: el dataset está completo.")
```

# 3.1. Transformación Crítica: Separación de Blood Pressure

```python
# 3.1. Separar 'Blood Pressure' en Sistólica y Diastólica
# La función astype(int) convierte los valores a números enteros.
df[['Systolic_BP', 'Diastolic_BP']] = df['Blood Pressure'].str.split('/', expand=True).astype(int)
df.drop('Blood Pressure', axis=1, inplace=True)

print("Columna 'Blood Pressure' separada y convertida a variables numéricas.")
```

# 3.2. Estandarización de Variables Categóricas

```python
# 3.2. Correcciones de Inconsistencias
df['Occupation'] = df['Occupation'].str.strip()
df['BMI Category'] = df['BMI Category'].str.strip()

# Corregir la inconsistencia 'Normal Weight' -> 'Normal'
df['BMI Category'] = df['BMI Category'].replace('Normal Weight', 'Normal')

print("Inconsistencias en variables categóricas corregidas.")

# 3.3. Conversión a Tipo 'Category' (Optimización)
categorical_cols = ['Gender', 'Occupation', 'Sleep Disorder', 'BMI Category']
for col in categorical_cols:
    df[col] = df[col].astype('category')

print("Variables categóricas convertidas a tipo 'category'.")
```

```python
# III. VALIDACIÓN FINAL

print("\n--- III. VALIDACIÓN FINAL ---")

# 1. Validación de tipos y nulos
print(df.info())

# 2. Vista final
print("\nVista previa del Dataset Final (Limpio y Transformado):")
print(df.head())

# 3. Conclusión sobre Multicolinealidad
print("\nNOTA IMPORTANTE para el Análisis:")
print("La fuerte correlación entre 'Quality of Sleep' y 'Stress Level' (r ≈ -0.91) es un *insight* valioso en sí mismo. No es necesario gestionarla, pero sí hay que mencionarla al interpretar estadísticas bivariadas.")

print("\n El conjunto de datos está en una condición óptima para el análisis estadístico y la generación de reportes.")
```
